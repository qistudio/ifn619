## Task 3: How many different species of kingfishers in Queensland?

Purpose: read a `json` file from a website; save a `json` file; navigate through a `json` file. 

The URL below is a subset of the Queensland wildlife database with information of family of "birds" in Queensland. 

Data source: [Qld WildNet Data API](https://www.data.qld.gov.au/dataset/qld-wildlife-data-api). 

You can use the [interactive documentation](https://wildnet-pub.science-data.qld.gov.au/docs#/) to navigate through the API to find the following information related to kingfishers:
- `kingdom_name`: Animalia
- `phylum_name`: Chordata
- `class_name`: Aves

In order to request species of kingfishers, we need to know the `family_name` of kingfishers first. Below, we use the request URL of all 'birds' in Queensland, and save the json file for Task 4. 

```python
bird_url = 'https://wildnet-pub.science-data.qld.gov.au/api/v1/families?class_name=Aves'

# Request the content from the web server
response = requests.get(bird_url)

# Preview the content
response.content
```

```python
# Get the data as json from the response
bird_family_json = response.json()
# Preview the data
bird_family_json
```

```python
# save the bird json file for later use in Task 4

# Dump the dict into a json string
qld_bird_json = json.dumps(bird_family_json)
# Write the json string to a file 
file_name = 'qld_bird_family.json'
with open(f"data/{file_name}","w") as fp:
    fp.write(qld_bird_json)
```

```python
# how many families of birds are in Queensland?
len(bird_family_json)
```

You will notice that it seems all the contents are stored as a list, each item is a dictionary that contains one family of birds. Where are the kingfishers?

**Question 1**: display the **6th** item from the list, and save the information in a new `json` file named `qld_kingfishers_family.json` under the `data/` folder. 

```python
# Display the 6th item
qld_kf_family = bird_family_json[5]
qld_kf_family
```

Now we know the `family_name` for kingfishers, we then can get the request URL to get all the kingfisher species in Queensland in order to answer the following questions: 
1. How many species of kingfishers are in Queensland?
2. What are the species names of kingfishers in Queensland?

```python
# Request URL (use Qld WildNet Data API documentation)
kingfishers_url = 'https://wildnet-pub.science-data.qld.gov.au/api/v1/species-list?family_name=Alcedinidae&page_size=1000&page_number=1'
kingfishers_url
```

```python
# Extract the information from the url above, save the data as json
response_kf = requests.get(kingfishers_url)
kf_json = response_kf.json()
kf_json
```

**Question 1**: How many species of kingfishers in Queensland?
>Hint: You can use `len([list])` to calculate how many items in a list. 

```python
# How many species of kingfishers in Queensland?
len(kf_json)
```

**Question 2**: What are the species names of kingfishers in Queensland?
>Hint: You need to use a for loop to extract the names

```python
# What are the species names (and common names) of kingfishers in Queensland?
for item in kf_json:
    print(item['scientific_name'], f"({item['accepted_common_name']})")
```
