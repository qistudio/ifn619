### Strategy in the context of uncertainty

One approach to understanding the complexity in environments is **Complex Dynamic Systems**

These have a number of characteristics, including:
- Involve large numbers of interacting elements that have some relationship with each other;
- The interactions are non-linear, and small changes can produce disproportionately big consequences;
- Dynamic and holistic: whole more than the sum of the parts, and solutions emerge from the unique combinations of elements;
- Co-evolutionary: the elements evolve with the environment, and with one another.


[Snowden (2005)](https://www.emerald.com/bss/article/6/1/47/100503/Strategy-in-the-context-of-uncertainty#idm46144037221456)

Complexity thinking provides a theoretical rooting for the common‐sense actions of strategists.

>"The dominant ideology in organizations assumes that in any “system” there are underlying relationships between cause and effect. These relationships can be discovered or approximated in such a way that the future can be planned on the basis of desired outcomes. Examples include the three‐year plan, the commission investigating past failure and scenario planning."


This approach to strategic thinking is to see the world not in terms of order versus chaos but in terms of various complexities.

One framework for modeling complex systems is the Cynefin Framework by Dave Snowden [Making Sense of Complexity - Cynefin Framework](https://cynefincentre.com/)

Maps varying levels of uncertainty across 4 domains, with varying levels of known & un-known information, as well as varying levels of what can be known, i.e. what is knowable & un-knowable.
Understanding the nature of the environment decisions are being made in, can support more effective decision making.


Data is not of itself informative without a shared context between giver and receiver, and an ability to communicate the significance in a meaningful way.
